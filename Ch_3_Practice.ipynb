3.1 How many seconds are in an hour? Use the interactive interpreter as a calculator and multiply the number of seconds in a minute (60) by the number of minutes in an hour (also 60).

```python
no_of_secs_in_mins = 60
no_of_mins_in_hr = 60
no_of_secs_in_hr = no_of_secs_in_mins * no_of_mins_in_hr
print(f"Number of seconds in an hour: {no_of_secs_in_hr}")
```

3.2 Assign the result from the previous task (seconds in an hour) to a variable called seconds_per_hour.

```python
seconds_per_hour = no_of_secs_in_hr
```

3.3 How many seconds are in a day? Use your seconds_per_hour variable.

```python
no_of_hrs_per_day = 24
print(f"Number of seconds per day: {no_of_hrs_per_day * seconds_per_hour}")
```

3.4 Calculate seconds per day again, but this time save the result in a variable called seconds_per_day.

```python
seconds_per_day = no_of_hrs_per_day * seconds_per_hour
print(f"Number of seconds per day: {seconds_per_day}")
```

3.5 Divide seconds_per_day by seconds_per_hour. Use floating-point (/) division.

```python
print(f"Number of hours in a day: {seconds_per_day / seconds_per_hour}")
```
